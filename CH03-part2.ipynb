```python
# 定义列表：存储医学检测核心特征名
lab_features = ["Glucose", "BMI", "BloodPressure", "Age", "Insulin"]
# 定义列表：存储常见糖尿病并发症
complications = ["糖尿病肾病", "糖尿病视网膜病变", "周围神经病变"]
# 列表索引取值（索引从0开始，取第一个检测特征）
first_feature = lab_features[0]
# 负索引取值（取最后一个并发症）
last_complication = complications[-1]
print("首个检测特征：", first_feature)
print("最后一个并发症：", last_complication)
# 列表常用操作：添加医学指标、修改病症名称
lab_features.append("HbA1c")  # 末尾添加糖化血红蛋白指标
complications[1] = "糖尿病眼底病变"  # 修改索引1的并发症名称
print("添加指标后的特征列表：", lab_features)
print("修改后的并发症列表：", complications)
```

```python
# 直接定义布尔值：标记检测结果是否异常
is_index_abnormal = True
is_patient_healthy = False

# 医学指标比较：判断血糖、血压是否超出正常范围
glucose = 165  # 血糖值(mg/dL)
bp = 95        # 舒张压(mmHg)
print("血糖高于正常阈值(140)：", glucose > 140)
print("舒张压等于正常阈值(80)：", bp == 80)
print("血糖不高于140：", not(glucose > 140))

# 逻辑运算组合：判断是否为高血糖合并高血压
is_diabetes_hypertension = (glucose > 140) and (bp > 90)
print("高血糖合并高血压：", is_diabetes_hypertension)

# 布尔值与数字转换：医学统计中量化判断结果
print("异常标记(True)转数字：", int(is_index_abnormal))
print("健康标记(False)转数字：", int(is_patient_healthy))
```

```python
import logging
import re
import pandas as pd

# --- 配置日志系统 ---
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('medical_data_processing.log', encoding='utf-8'),
        logging.StreamHandler()
    ]
)

logger = logging.getLogger('DiabetesProcessor')

# 模拟一组存在问题的患者数据
problematic_patients = [
    {"id": "P001", "glucose": "150mg/dL"},  # 含单位字符串
    {"id": "P002", "glucose": None},       # 缺失值
    {"id": "P003", "glucose": -5},         # 生理异常值
    {"id": "P004", "glucose": 620}         # 极端异常值
]

def clean_glucose_value(raw_value):
    """
    清洗血糖数据：
    1. 去除单位
    2. 转换为 float
    3. 返回标准数值
    """
    if raw_value is None:
        return None

    if isinstance(raw_value, str):
        # 提取字符串中的数字部分（支持小数）
        match = re.search(r"[-+]?\d*\.?\d+", raw_value)
        if match:
            return float(match.group())
        else:
            raise ValueError("无法从字符串中提取数值")

    return float(raw_value)


for patient in problematic_patients:
    pid = patient.get("id", "Unknown")

    try:
        # 数据清洗
        glucose_value = clean_glucose_value(patient.get("glucose"))

        # 缺失值校验
        if pd.isna(glucose_value):
            raise ValueError("血糖值为缺失值（NaN）")

        # 生理逻辑校验
        if glucose_value < 0:
            raise ValueError(f"血糖值{glucose_value}不符合生理常识")

        # 医学合理范围校验（例如 0~600 mg/dL）
        if not (0 <= glucose_value <= 600):
            raise ValueError(f"血糖值{glucose_value}超出医学合理范围")

        logger.info(f"患者{pid}血糖值校验通过：{glucose_value} mg/dL")

    except ValueError as e:
        logger.warning(f"患者{pid}数据异常：{e}，已记录并跳过")

    except Exception as e:
        logger.critical(f"患者{pid}发生系统级错误：{e}")
        # 不再终止程序，而是继续处理下一个患者
        continue
```
